```python
import subprocess
import os
import re
from pathlib import Path

os.chdir('/jupyter/dl3/dl3')
PROJECT_ROOT = Path('/jupyter/dl3/dl3')

# Copy current environment and add your custom variables
env = os.environ.copy()
env['PYTHONPATH'] = '/jupyter/dl3/dl3:' + env.get('PYTHONPATH', '')
env['DATASET_ROOT'] = str(PROJECT_ROOT / 'dataset')
env['OUTPUT_ROOT'] = str(PROJECT_ROOT / 'analysis_outputs')
env['TRAIN_IMAGES_DIR'] = str(PROJECT_ROOT / 'dataset' / 'train' / 'images_train')
env['TRAIN_LABELS_CSV'] = str(PROJECT_ROOT / 'dataset' / 'train' / 'labels_train.csv')
env['VAL_IMAGES_DIR'] = str(PROJECT_ROOT / 'dataset' / 'val' / 'images_val')
env['VAL_LABELS_CSV'] = str(PROJECT_ROOT / 'dataset' / 'val' / 'labels_val.csv')
env['TEST_IMAGES_DIR'] = str(PROJECT_ROOT / 'dataset' / 'test' / 'images')
env['TEST_MANIFEST_CSV'] = str(PROJECT_ROOT / 'dataset' / 'test' / 'manifest_public.csv')

process = subprocess.Popen(
    ['python', '-m', 'src.models.train', 
     '--model', 'vit_b16',
     '--epochs', '30',
     '--batch-size', '32',
     '--lr', '3e-4',
     '--weight-decay', '5e-2',
     '--step-size', '20',
     '--gamma', '0.1',
     '--num-workers', '1'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
    env=env  # Pass the environment variables!
)

last_percent = -1
for line in process.stdout:
    match = re.search(r'Train:\s+(\d+)%', line)
    if match:
        percent = int(match.group(1))
        if percent % 10 == 0 and percent != last_percent:
            print(line, end='')
            last_percent = percent
    else:
        print(line, end='')
```
